## Task 1.1 — Your first notebook
Generate and plot the unit impulse, unit step, unit ramp and real exponential sequences.

````python

n = np.arange(-5, 11)
delta = (n == 0).astype(float)       # unit impulse
step = (n >= 0).astype(float)        # unit step
ramp = n * step                     # unit ramp
expo = 0.8**n * step                # real exponential

fig, axes = plt.subplots(1, 4, figsize=(13, 2.8))
for ax, x, title in zip(
    axes, [delta, step, ramp, expo],
    ["Unit impulse δ(n)", "Unit step u(n)", "Unit ramp r(n)", "(0.8)^n u(n)"]
):
    ax.stem(n, x, basefmt="k-")
    ax.set_title(title)
    ax.set_xlabel("n")
    ax.set_ylabel("Amplitude")
plt.tight_layout()
plt.show()
````

### Observations

1. **δ(n):** Produces a single spike at n = 0.
2. **u(n):** Has a value of 0 for n < 0 and 1 for n ≥ 0.
3. **r(n):** Has a value of 0 for n < 0 and increases as 0, 1, 2, 3, …
4. **(0.8)ⁿu(n):** Begins at 1 when n = 0 and gradually decreases towards zero.

## Task 1.2 — Change the parameters


````python
import numpy as np
import matplotlib.pyplot as plt
n = np.arange(-5, 11)
u = (n >= 0).astype(float)
fig, axes = plt.subplots(1, 3, figsize=(12, 3))
for ax, a in zip(axes, [0.5, -0.8, 1.1]):
    y = (a**n) * u
    ax.stem(n, y, basefmt="k-")
    ax.set_title(f"a = {a}")
    ax.set_xlabel("n")
    ax.set_ylabel("Amplitude")
plt.tight_layout()
plt.show()

n = np.arange(0, 41)
plt.figure(figsize=(10, 4))
plt.stem(n, np.cos(2*np.pi*n/8), linefmt="C0-", markerfmt="C0o",
         basefmt=" ", label="cos(2πn/8)")
plt.plot(n, np.cos(2*np.pi*n/7.5), "r.--", label="cos(2πn/7.5)")
plt.xlabel("Sample index n")
plt.ylabel("Amplitude")
plt.title("Comparison of two discrete-time cosine sequences")
plt.legend()
plt.grid(True)
plt.show()
````

**Answer:** `cos(2πn/8)` repeats exactly. Its period is 8 samples because the angular frequency is 2π/8. `cos(2πn/7.5)` does not repeat exactly at integer sample indices because its period would be 7.5 samples, which is not an integer. A discrete-time sinusoid is periodic only when its period in samples is an integer (equivalently, its frequency ratio is rational).

## Task 1.3 — Working with audio



````python
fname = "Audio.wav"
import numpy as np
import matplotlib.pyplot as plt
if os.path.exists(fname):
    fs, x = wavfile.read(fname)
    if x.ndim > 1:                  # stereo to mono
        x = x.mean(axis=1)
    x = x.astype(float)
    peak = np.max(np.abs(x))
    if peak > 0:
        x = x / peak                # normalise to [-1, 1]
    audio_source = "Your voice recording"
else:
    fs = 16000
    t_fallback = np.arange(0, 2, 1/fs)
    x = 0.5*np.sin(2*np.pi*440*t_fallback) * np.exp(-t_fallback)
    audio_source = "Fallback 440 Hz test signal"

t = np.arange(len(x)) / fs
print(f"Audio source: {audio_source}")
print(f"Sampling rate = {fs} Hz")
print(f"Duration = {len(x)/fs:.2f} s")
print(f"Number of samples = {len(x)}")

plt.figure(figsize=(10, 3.5))
plt.plot(t, x)
plt.xlabel("Time (s)")
plt.ylabel("Amplitude")
plt.title(audio_source)
plt.grid(True)
plt.show()
````

### Difference Observed between Python and Octave Syntax Comparison

|**Python** | **Octave/MATLAB** |
|---|---|
| `np.arange(-5, 11)` | `n = -5:10;` |
| `(n == 0).astype(float)` | `double(n == 0)` |
| `(n >= 0).astype(float)` | `double(n >= 0)` |
| `n * step` | `n .* step` |
| `0.8**n * step` | `(0.8 .^ n) .* step` |
| `plt.subplots(1,4)` | `subplot(1,4,k)` |
| `ax.stem(..., basefmt="k-")` | `stem(n, x, 'filled')` |
| `plt.tight_layout()` | Not needed |
| `plt.show()` | Automatic in Octave |

## Analysis questions

### 1. What is the sampling rate of your phone’s recording? How many numbers are needed to store one minute of it in mono?

The sampling rate is the value printed by the audio cell above. For a one-minute mono recording, the number of samples is:

**Number of samples = sampling rate × 60.**

For example, at 44,100 Hz: 44,100 × 60 = **2,646,000 samples**. At 48,000 Hz: 48,000 × 60 = **2,880,000 samples**. The exact answer for your recording depends on its printed sampling rate.

### 2. Why is a stem plot more appropriate than a continuous line for x(n)?

A discrete-time signal is defined only at integer sample indices. A stem plot shows the individual sample values and makes it clear that no values are defined between the samples. A continuous line can misleadingly suggest that the signal exists at every time instant.

### 3. For a = 1.1 the exponential grows. Which property of a system would such a signal violate if it were an impulse response?

It would violate **BIBO stability** if it were an impulse response. For a discrete-time LTI system, stability requires the impulse response to be absolutely summable: the sum of the absolute values of all impulse-response samples must be finite. The sequence 1.1ⁿu(n) grows and its absolute sum diverges, so the system would be unstable.
